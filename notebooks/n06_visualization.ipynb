```python
import matplotlib.pyplot as plt

models = ["UNet + Aug", "UNet", "Random Forest"]
training_time = [18.83, 20.05, 3.32]   # minutes
model_size = [118.5, 118.5, 504.5]     # MB

fig, ax1 = plt.subplots(figsize=(6,4))

# Bar chart for training time
ax1.bar(models, training_time)
ax1.set_ylabel("Training Time (minutes)")
ax1.set_xlabel("Model")

# Second axis for model size
ax2 = ax1.twinx()
ax2.plot(models, model_size, marker='o')
ax2.set_ylabel("Model Size (MB)")

plt.title("Model Comparison")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

models = ["UNet + Aug", "UNet", "Random Forest"]
training_time = [18.83, 20.05, 3.32]
model_size = [118.5, 118.5, 504.5]

x = np.arange(len(models))
width = 0.35

plt.figure(figsize=(6,4))
plt.bar(x - width/2, training_time, width, label="Training Time (min)")
plt.bar(x + width/2, model_size, width, label="Model Size (MB)")

plt.xticks(x, models)
plt.legend()
plt.title("Model Comparison")
plt.show()
```

```python
import matplotlib.pyplot as plt

models = ["UNet + Aug", "UNet", "Random Forest"]
training_time = [18.83, 20.05, 3.32]   # minutes
model_size = [118.5, 118.5, 504.5]     # MB

best_model = "UNet + Aug"

# Colors (highlight best model in red)
colors_time = ["gray", "gray", "tab:blue"]
colors_size = ["tab:blue", "tab:blue", "gray"]

fig, axes = plt.subplots(2, 1, figsize=(5,5))  # smaller height

# --- Training Time ---
axes[0].bar(models, training_time, color=colors_time)
axes[0].set_ylabel("Training Time (min)")
axes[0].set_title("Training Time Comparison")
axes[0].set_ylim(0, max(training_time) * 1.2)

# Add values on bars
for i, v in enumerate(training_time):
    axes[0].text(i, v + 0.5, f"{v:.1f}", ha='center')

# --- Model Size ---
axes[1].bar(models, model_size, color=colors_size)
axes[1].set_ylabel("Model Size (MB)")
axes[1].set_title("Model Size Comparison")
axes[1].set_ylim(0, max(model_size) * 1.2)

# Add values on bars
for i, v in enumerate(model_size):
    axes[1].text(i, v + 10, f"{v:.1f}", ha='center')

plt.tight_layout()
plt.show()
```
